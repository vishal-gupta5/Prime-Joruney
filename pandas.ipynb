```python
#Cleaning Data - Missing Values
df = pd.read_csv("raw_data.csv")
df.isnull() #It returns "True" in-place of "NaN"
df.isnull().sum() #Count the missing value of each columns
df.dropna() #Drops all the null values from our Dataset as Drops rows wise
df.dropna(axis=1) #Drops all the null values from our Dataset as Drops column wise
df.fillna(0) #Fill the particular value in-place of NaN

age_mean = df["age"].mean()
df["age"].fillna(age_mean)

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean();
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean);
# cleaned_data

df.ffill # Forward fill
df.bfill() # Backword fill
```

```python
#Cleaing Data --> Handle De-Duplication
df2 = df.copy()
df2.duplicated()
df2["country"].duplicated()
df2[["country", "gender"]].duplicated()
df2.drop_duplicates(inplace=True)
df2
```
