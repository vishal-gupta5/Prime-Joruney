```python
import pandas as pd 
```

```python
#Acess through Index
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(s[0])
print(s[2])
print(type(s))
```

```python
#Access through Label 
s = pd.Series([1, 2, 3, 4, 5], index=["Adam", "eve", "Bob", "Alice", "Carlie"])
print(s)
print(s["Bob"])
print(s.index)
```

```python
#Vectorized Operation
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([10, 20, 30, 40, 50])

print (s1 + s2)
s1[0] = 100 #Modify the array
print(s1)

changed_s1 = s1.drop(0) #Drop the value from an Array
print(changed_s1)
```

```python
#DataFrame (Object)

info = {
    "Name": ["Adam", "eve", "Bob", "Alice", "Carlie"],
    "Age": [22, 23, 24, 25, 26],
    "Cgpa": [8.9, 9.2, 7.8, 9.7, 8.5]
}

df = pd.DataFrame(info)
print(df, type(df))
print(df.index)
print(df.columns)
```

```python
#DataFrame (List of Lists)

df = pd.DataFrame([["Vishal", 25], ["Himanshu", 22], ["Nikhil", 18], ["Arjun", 15]], columns=["Name", "Age"])
print(df)
```

```python
#DataFrame (Numpy Array)
import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]]);
df = pd.DataFrame(np_arr, columns=["A", "B", "C"]);
print(df)
```
