```python
# Real Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]

years = [2008, 2009, 2010, 2011, 2012]
plt.plot(years, oscar_revenue, color="red", marker="o", linestyle="--") # Line Plot
plt.title("Oscar Movies revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenues (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, "<-y", label="Non-Oscar Movies")
plt.title("Oscar movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
```

```python
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.style.available
plt.grid()
plt.style.use("default")
plt.show()
```

```python
#Vertical Bar Chart

movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

 
plt.bar(years, oscar_revenue, color="green")
plt.title("Revenue in each Year for oscar")
plt.xlabel("Years")
plt.ylabel("Revenue (In $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i] + 20, str(oscar_revenue[i]), ha="center")

plt.ylim(0, max(oscar_revenue) + 100)
```

```python
import numpy as np
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

x = np.arange(len(years))
width = 0.4

plt.bar(x - width / 2, oscar_revenue, width, label="Oscar Movies") #Shifting to left
plt.bar(x + width / 2, non_oscar_revenue, width, label="Non-Oscar Movies") #Shifting to right

plt.title("Oscar vs Non-Oscar Movie Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (In $M)")
plt.legend()
plt.xticks(x, years)
```

```python
#Horizontal Bar chart

oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]

plt.barh(oscar_movies, oscar_revenue)
plt.xlabel("Oscar Revenue")
plt.ylabel("Oscar Movies")
```
