```python
# Real Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]

years = [2008, 2009, 2010, 2011, 2012]
plt.plot(years, oscar_revenue, color="red", marker="o", linestyle="--") # Line Plot
plt.title("Oscar Movies revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenues (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, "<-y", label="Non-Oscar Movies")
plt.title("Oscar movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
```

```python
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.style.available
plt.grid()
plt.style.use("default")
plt.show()
```

```python
#Vertical Bar Chart

movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

 
plt.bar(years, oscar_revenue, color="green")
plt.title("Revenue in each Year for oscar")
plt.xlabel("Years")
plt.ylabel("Revenue (In $M)")

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i] + 20, str(oscar_revenue[i]), ha="center")

plt.ylim(0, max(oscar_revenue) + 100)
```

```python
import numpy as np
oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

x = np.arange(len(years))
width = 0.4

plt.bar(x - width / 2, oscar_revenue, width, label="Oscar Movies") #Shifting to left
plt.bar(x + width / 2, non_oscar_revenue, width, label="Non-Oscar Movies") #Shifting to right

plt.title("Oscar vs Non-Oscar Movie Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (In $M)")
plt.legend()
plt.xticks(x, years)
```

```python
#Horizontal Bar chart

oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]

plt.barh(oscar_movies, oscar_revenue)
plt.xlabel("Oscar Revenue")
plt.ylabel("Oscar Movies")
```

```python
# Scatter Plots
people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]

plt.scatter(age, bp, s=bp, cmap="plasma", c=bp)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i] + 2, bp[i]))

plt.xlim(min(age), max(age) + 10)
plt.ylim(min(bp), max(bp) + 5)
```

```python
# Multiple scatter plots
cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (°C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (°C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp, winter_humidity, label="Winter")
plt.scatter(summer_temp, summer_humidity, label="Summer")
plt.title("Summer vs Winter Data")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

```python
#Pie Chart
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explode = [0, 0, 0, 0.1, 0]



plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
           wedgeprops = {
                "edgecolor": "black",
               "linewidth": 1,
               "linestyle": "-"
           },
        explode = explode,
        startangle = 45
       )

plt.show()
```
